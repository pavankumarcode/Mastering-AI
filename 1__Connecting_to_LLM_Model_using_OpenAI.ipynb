```python
import os
from openai import OpenAI
#import anthropic
import google.genai as genai
from google.colab import userdata
```

```python
# 1. OPENAI (GPT-4o, GPT-4o-mini)
from typing import Optional

def query_openai(
    prompt: str,
    system_prompt: str = "You are a helpful assistant.",
    model: str = "gpt-4o-mini",
    temperature: float = 0.7
) -> str:
    """
    Connects to OpenAI's official API using best practices for SWEs.

    Foundations of AI:
    - Model Selection: 'gpt-4o-mini' is optimized for cost and speed, ideal for simple logic.
    - Temperature: Controls randomness. 0.0 is deterministic (good for code/extraction),
      while 1.0+ is creative (good for brainstorming).
    - Message Roles:
        - 'system': Defines the persona and constraints.
        - 'user': The specific instruction.
    """
    print(f"\n--- Querying OpenAI ({model}) ---")

    # Best Practice: Initialize client once or use a singleton in production.
    # For this notebook, we fetch the key dynamically.
    api_key = userdata.get('OPENAI_API_KEY')
    if not api_key:
        return "Error: OPENAI_API_KEY not found in Colab Secrets."

    client = OpenAI(api_key=api_key)

    try:
        # Note: Using modern parameters (max_completion_tokens vs max_tokens)
        response = client.chat.completions.create(
            model=model,
            messages=[
                {"role": "system", "content": system_prompt},
                {"role": "user", "content": prompt}
            ],
            temperature=temperature,
        )

        # Best Practice: Safely access nested attributes
        content = response.choices[0].message.content
        return content if content else "Error: Received empty response from model."

    except Exception as e:
        # Best Practice: In a real app, log the full traceback but return a clean error
        return f"OpenAI API Error: {str(e)}"
```

```python
resposnse_text = query_openai(prompt='Tell me about somthing about today, in less than 10 words.')
print(f'Response Received: [{resposnse_text}]')
```
